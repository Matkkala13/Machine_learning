Classifying model reviews (positive and negative): Binary classification example

The IMDb dataset: 50000 polarized reviews from the Internet Movie Database

25000 reviews for training and 25000 for testing, each one containing 50% positive reviews and 50% negative.

It has been preprocessed so that each review is expressed as a sequence of integers (words). Each integer is a specific word in the dictionary.

Loading the IMDb dataset

```python
from keras.datasets import imdb

(train_data, train_labels), (test_data, test_labels) = imdb.load_data(
    num_words=10000
)
```

```python
train_data[0], train_labels[0]
```

How to check the reviews: 

```python
word_index = imdb.get_word_index()
reverse_word_index = dict([(value,key) for (key,value) in word_index.items()])
decoded_review = " ".join(
    [reverse_word_index.get(i - 3, "?") for i in train_data[0]]
)
decoded_review
```

Preparing the data

Vectorizing input sequences and labels

```python
import numpy as np

def multi_hot_encode(sequences, num_classes):
    results = np.zeros((len(sequences), num_classes))
    for i, sequence in enumerate(sequences):
        results[i][sequence] = 1.0
    return results

x_train = multi_hot_encode(train_data, num_classes=10000)
x_test = multi_hot_encode(test_data, num_classes=10000)

x_train[0]
```

```python
y_train = train_labels.astype("float32")
y_test = test_labels.astype("float32")

y_train[0]
```

Building the model

16 units in the first two neurons, the last one is 1 because it uses sigmoid activation, meaning the answer the model gives would be between 0 and 1, if its closer to 0 it will be 0 and if it is closer to 1 it will be 1, matching the problem of binary classification (positive and negative reviews)

```python
import keras 
from keras import layers

model = keras.Sequential(
    [
        layers.Dense(16, activation="relu"),
        layers.Dense(16, activation="relu"),
        layers.Dense(1, activation="sigmoid")
    ]
)
```

Compiling the model

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
```

Validation set

```python
x_val = x_train[:10000]
partial_x_train = x_train[10000:]
y_val = y_train[:10000]
partial_y_train = y_train[10000:]
```

Training

validation data is made for measuring loss and accuracy with inputs that arent on training (so they are new to the model and its a more truthful measure)

```python
history = model.fit(
    partial_x_train, 
    partial_y_train,
    epochs=20,
    batch_size=512,
    validation_data=(x_val, y_val),
)
```

an alternative for using validation data is using the validation_split parameter 
and specifying the fraction of the data to be used as validation
